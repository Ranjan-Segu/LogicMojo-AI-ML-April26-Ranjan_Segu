**THE PRINT SYNTAX**

This is one of the most important concepts in Python because print() has default behavior, and the end parameter changes that behavior.

Understanding print('*', end='')

Let's start with the simplest print statement.

```python
print('*')
```

Most beginners think Python only prints *.

But that's not exactly what happens.

Python actually prints:

```python
*
(new line)
```

The new line is invisible.

So Python is really doing something like this:

Print *

↓

Move the cursor to the beginning of the next line

What is end?

The print() function has a parameter called end.

Its job is to tell Python:

"What should I print after printing the value?"

The default value is

end='\n'

So these two statements are equivalent:

print('*')

and

print('*', end='\n')

Both produce

*
followed by moving to the next line.

**Python [] (Square Brackets) Usage by Data Type**

---



| Data Type                   | Usage                          | Syntax                        | Example                     |
| --------------------------- | ------------------------------ | ----------------------------- | --------------------------- |
| **List (`list`)**           | Create an empty list           | `[]`                          | `l = []`                    |
|                             | Create a list                  | `[item1, item2, ...]`         | `l = [1, 2, 3]`             |
|                             | Access an element (indexing)   | `list[index]`                 | `l[0]`                      |
|                             | Access a portion (slicing)     | `list[start:stop:step]`       | `l[1:4]`, `l[::-1]`         |
|                             | Modify an element              | `list[index] = value`         | `l[2] = 100`                |
|                             | Replace multiple elements      | `list[start:stop] = iterable` | `l[1:3] = [10, 20]`         |
| **String (`str`)**          | Access a character (indexing)  | `string[index]`               | `s[0]`                      |
|                             | Access a portion (slicing)     | `string[start:stop:step]`     | `s[1:4]`, `s[::-1]`         |
| **Tuple (`tuple`)**         | Access an element (indexing)   | `tuple[index]`                | `t[1]`                      |
|                             | Access a portion (slicing)     | `tuple[start:stop:step]`      | `t[1:]`, `t[::-1]`          |
| **Dictionary (`dict`)**     | Access a value using a key     | `dict[key]`                   | `d["name"]`                 |
|                             | Add or update a key-value pair | `dict[key] = value`           | `d["age"] = 30`             |
|                             | Delete a key-value pair        | `del dict[key]`               | `del d["age"]`              |
| **Bytes (`bytes`)**         | Access a byte (indexing)       | `bytes[index]`                | `b[0]`                      |
|                             | Access a portion (slicing)     | `bytes[start:stop:step]`      | `b[1:4]`                    |
| **Bytearray (`bytearray`)** | Access a byte (indexing)       | `bytearray[index]`            | `ba[0]`                     |
|                             | Access a portion (slicing)     | `bytearray[start:stop:step]`  | `ba[2:5]`                   |
|                             | Modify a byte                  | `bytearray[index] = value`    | `ba[0] = 65`                |
| **Set (`set`)**             | Not supported                  | —                             | `s[0]` → ❌ `TypeError`      |
| **Boolean (`bool`)**        | Not supported                  | —                             | `True[0]` → ❌ `TypeError`   |
| **Integer (`int`)**         | Not supported                  | —                             | `123[0]` → ❌ `TypeError`    |
| **Float (`float`)**         | Not supported                  | —                             | `45.6[0]` → ❌ `TypeError`   |
| **Complex (`complex`)**     | Not supported                  | —                             | `(3+4j)[0]` → ❌ `TypeError` |
